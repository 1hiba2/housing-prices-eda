```python
df = pd.read_csv("Housing.csv")
df.head()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="price",
    bins=30,
    kde=True,
    color="skyblue"
)

plt.title("Distribution des prix des maisons")
plt.xlabel("Prix des maisons")
plt.ylabel("Fréquence")

plt.show()
```


La distribution des prix des maisons présente une asymétrie à droite la majorité des maisons possède des prix relativement faibles ou moyens tandis qu'un nombre limité de maisons présente des prix très élevés cela indique que les prix immobiliers ne suivent pas une distribution normale et que certaines maisons constituent des valeurs extrêmes.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="area",
    bins=30,
    kde=True,
    color="lightgreen"
)

plt.title("Distribution de la superficie des maisons")
plt.xlabel("Superficie des maisons")
plt.ylabel("Fréquence")

plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.countplot(data=df, x="bedrooms", color="skyblue")
plt.title("Distribution du nombre de chambres")
plt.xlabel("Nombre de chambres")
plt.ylabel("Nombre de maisons")
plt.subplot(1, 2, 2)
sns.countplot(data=df, x="bathrooms", color="lightgreen")
plt.title("Distribution du nombre de salles de bains")
plt.xlabel("Nombre de salles de bains")
plt.ylabel("Nombre de maisons")

plt.tight_layout()
plt.show()
```

La majorité des maisons possède 3 chambres et une seule salle de bains. Les maisons avec davantage de chambres ou de salles de bains sont moins fréquentes.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x="area", y="price", color="blue")

plt.title("Relation entre la superficie et le prix")
plt.xlabel("Superficie")
plt.ylabel("Prix")

plt.show()
```

On observe une corrélation positive entre la superficie et le prix. Plus la superficie augmente, plus le prix des maisons a tendance à augmenter

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="bedrooms",
    y="price",
    color="lightblue"
)

plt.title("Relation entre le nombre de chambres et le prix")
plt.xlabel("Nombre de chambres")
plt.ylabel("Prix des maisons")

plt.show()
```

Le prix des maisons a généralement tendance à augmenter avec le nombre de chambres. Cependant, cette relation n'est pas toujours régulière

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="airconditioning",
    y="price",
    color="lightgreen"
)

plt.title("Relation entre la climatisation et le prix")
plt.xlabel("Climatisation")
plt.ylabel("Prix des maisons")

plt.show()
```

Les maisons avec climatisation ont généralement des prix plus élevés que celles sans climatisation

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="furnishingstatus",
    y="price",
    color="lightpink"
)

plt.title("Impact de l'ameublement sur le prix")
plt.xlabel("Statut d'ameublement")
plt.ylabel("Prix des maisons")

plt.show()
```

Les maisons meublées sont généralement plus chères que les maisons semi meublées et non meublées

```python
df_numeric = df.select_dtypes(include=["number"])

correlation_matrix = df_numeric.corr()
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matrice de corrélation")
plt.show()
```

```python
correlation_price = df.select_dtypes(
    include=["number"]
).corr()["price"].drop("price")


correlation_price = correlation_price.sort_values(
    ascending=False
)

print(correlation_price)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=correlation_price.values,
    y=correlation_price.index,
    color="skyblue"
)

plt.title("Variables les plus corrélées avec le prix")
plt.xlabel("Coefficient de corrélation")
plt.ylabel("Variables")

plt.show()
```
